```python
from PIL import Image       #small 'from'
```

```python
img_pil = Image.open('/Users/simarkalsi/ANALYSIS/TomatoDataset/augmented_split_dataset/train/Tomato_Late_blight/img_2.jpg')
img_np = np.array(img_pil)

print("Image shape using PIL:", img_np.shape)
print("Image shape using PIL:", img_np.dtype)
print("Image shape using PIL:", type(img_np))
```

```python
red_channel = img_np[:, :, 0]
green_channel = img_np[:, :, 1]
blue_channel = img_np[:, :, 2]  

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(red_channel, cmap='Reds')
axes[0].set_title('Red Channel')
axes[0].axis('off')

axes[1].imshow(green_channel, cmap='Greens')
axes[1].set_title('Green Channel')
axes[1].axis('off')

axes[2].imshow(blue_channel, cmap='Blues')  
axes[2].set_title('Blue Channel')
axes[2].axis('off') 

plt.show()
```

The background is dark and strong in all three channels. That means it has high R, G, and B values. High values in all three channels together make a bright, near-white or grey colour. So your photo was taken on a bright, plain background, which is typical of lab-style datasets.

```python
print(img_np[10, 10])     # a background pixel (corner)
print(img_np[128, 128])   # a pixel on the leaf (centre)
```

```python

for cls in sorted(os.listdir(train_dir)):
    if cls == ".DS_Store":
        continue
    folder = train_dir + "/" + cls
    print(cls, len(os.listdir(folder)))
```

```python
for cls in sorted(os.listdir(train_dir)):
    if cls == ".DS_Store":
        continue
    for f in os.listdir(train_dir + "/" + cls):
        if not f.lower().endswith((".jpg", ".jpeg", ".png")):
            print(cls, f)
```

```python

for cls in sorted(os.listdir(train_dir)):
    if cls == ".DS_Store":
        continue
    folder = train_dir + "/" + cls
    images = [f for f in os.listdir(folder) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    print(cls, len(images))
```

```python

for cls in sorted(os.listdir(test_dir)):
    if cls == ".DS_Store":
        continue
    folder = test_dir + "/" + cls
    images = [f for f in os.listdir(folder) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    print(cls, len(images))
```

```python

for cls in sorted(os.listdir(val_dir)):
    if cls == ".DS_Store":
        continue
    folder = val_dir + "/" + cls
    images = [f for f in os.listdir(folder) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    print(cls, len(images))
```

```python
classes = []
counts = []
```

```python
for cls in sorted(os.listdir(train_dir)):
    if cls.startswith("."):
        continue
    images = [f for f in os.listdir(train_dir + "/" + cls) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    classes.append(cls)
    counts.append(len(images))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.bar(classes, counts, color='skyblue')
plt.xticks(rotation=45, ha='right')
plt.title('Number of Images per Class in Training Set')
plt.show()
```
