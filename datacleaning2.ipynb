```python
import os           #works with folders and files ( listing them, building paths)
from PIL import Image   #Image efrom PIL (Python Imaging Library) to open and manipulate images
from collections import Counter     # counter is a ready-made tally sheet that counts the number of occurrences of each item in a list

train_dir = '/Users/simarkalsi/ANALYSIS/TomatoDataset/augmented_split_dataset/train'

sizes = Counter()   
modes = Counter()
broken = []

for cls in sorted(os.listdir(train_dir)):
    if cls.startswith("."):     #for .DSStore files
        continue                #skip those
    for f in os.listdir(train_dir + "/" + cls):
        if not f.lower().endswith((".jpg", ".jpeg", ".png")):
            continue
        path = train_dir + "/" + cls + "/" + f
        try:
            img = Image.open(path)
            sizes[img.size] += 1
            modes[img.mode] += 1
        except Exception:
            broken.append(path)

print("Sizes:", sizes)
print("Modes:", modes)
print("Broken files:", len(broken))
```
